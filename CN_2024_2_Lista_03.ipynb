### Corrida de Newton

```python
def sistemaEquacoes(x, y):
    f1 = 3*x**3 + 2*y**2 - x*y + 16*y + 3
    f2 = 2*x**3 - 3*y + 11*x
    return np.array([f1, f2])

def jacobiana(x, y):
    df1dx = 9*x**2 - y
    df1dy = 4*y - x + 16
    df2dx = 6*x**2 + 11
    df2dy = -3
    return np.array([[df1dx, df1dy], [df2dx, df2dy]])

def metodoNewton(x0, y0, tolerancia=1e-6, maxIter=10):
    x, y = x0, y0
    print(f"Iteração 0: x = {x:.1f}, y = {y:.1f}")

    for i in range(1, maxIter + 1):
        J = jacobiana(x, y)
        F = sistemaEquacoes(x, y)
        delta = np.linalg.solve(J, -F)
        x += delta[0]
        y += delta[1]

        print(f"Iteração {i}: x = {x:.6f}, y = {y:.6f}")

        if np.linalg.norm(delta, ord=np.inf) < tolerancia:
            break

    print(f"\nSolução final encontrada pelo Método de Newton: x = {x:.6f}, y = {y:.6f}")
    print(f"Número total de iterações: {i}\n")
    return i, x, y

def metodoNewtonModificado(x0, y0, tolerancia=1e-6, maxIter=10):
    x, y = x0, y0
    JNewtonModificado = jacobiana(x, y)

    print(f"Iteração 0: x = {x:.1f}, y = {y:.1f}")

    for i in range(1, maxIter + 1):
        F = sistemaEquacoes(x, y)
        delta = np.linalg.solve(JNewtonModificado, -F)
        x += delta[0]
        y += delta[1]

        print(f"Iteração {i}: x = {x:.6f}, y = {y:.6f}")

        if np.linalg.norm(delta, ord=np.inf) < tolerancia:
            break

    print(f"\nSolução final encontrada pelo Método de Newton Modificado: x = {x:.6f}, y = {y:.6f}")
    print(f"Número total de iterações: {i}\n")
    return i, x, y

def resultado(x0, y0):
    iteracoesNewton, xNewton, yNewton = metodoNewton(x0, y0)
    iteracoesNewtonModificado, xNewtonModificado, yNewtonModificado = metodoNewtonModificado(x0, y0)

    if iteracoesNewton < iteracoesNewtonModificado:
        print("O método de Newton convergiu mais rápido.")
    elif iteracoesNewton > iteracoesNewtonModificado:
        print("O método de Newton Modificado convergiu mais rápido.")
    else:
        print("Ambos os métodos convergiram no mesmo número de iterações.")

x0 = float(input())
y0 = float(input())

resultado(x0, y0)
```

### O desafio imposto a Runge-Kutta

```python
def funcao(x, y):
    return x**2 + y

def solucaoExata(x):
    return -x**2 - 2*x - 2 + 2 * np.exp(x)

def rungeKuttaSegundaOrdem(x0, y0, h, n):
    x, y = x0, y0

    for _ in range(n):
        k1 = funcao(x, y)
        k2 = funcao(x + h, y + h * k1)
        y += (h / 2) * (k1 + k2)
        x += h

    return y

def rungeKuttaQuartaOrdem(x0, y0, h, n):
    x, y = x0, y0

    for _ in range(n):
        k1 = funcao(x, y)
        k2 = funcao(x + h / 2, y + (h / 2) * k1)
        k3 = funcao(x + h / 2, y + (h / 2) * k2)
        k4 = funcao(x + h, y + h * k3)
        y += (h / 6) * (k1 + 2*k2 + 2*k3 + k4)
        x += h

    return y

def erroRelativo(aproximado, exato):
    return abs((exato - aproximado) / exato) * 100

def resultado(xFinal, x0=0, y0=0, n = 4):
  h = xFinal/n

  resultado = solucaoExata(xFinal)
  rk2 = rungeKuttaSegundaOrdem(x0, y0, h, n)
  rk4 = rungeKuttaQuartaOrdem(x0, y0, h, n)

  erroRk2 = erroRelativo(rk2, resultado)
  erroRk4 = erroRelativo(rk4, resultado)

  print(f"A solução exata é: y({xFinal:.1f}) = {resultado:.4f}")
  print(f"A solução pelo método de Runge-Kutta de Segunda Ordem é: y({xFinal:.1f}) = {rk2:.4f}")
  print(f"O erro relativo pelo método de Runge-Kutta de Segunda Ordem é de {erroRk2:.5f}%")
  print(f"A solução pelo método de Runge-Kutta é: y({xFinal:.1f}) = {rk4:.4f}")
  print(f"O erro relativo pelo método de Runge-Kutta é de {erroRk4:.5f}%")
  print("O método de Runge-Kutta de Quarta Ordem é mais eficiente" if erroRk4 < erroRk2 else "O método de Runge-Kutta de Segunda Ordem é mais eficiente")

xFinal = float(input())
resultado(xFinal)
```

### Peter e os dados desajustados.

```python
def calcularRetaMmq(n, xArray, yArray):
    somaX = sum(xArray)
    somaY = sum(yArray)
    somaXy = sum(x * y for x, y in zip(xArray, yArray))
    somaXx = sum(x * x for x in xArray)

    a = (n * somaXy - somaX * somaY) / (n * somaXx - somaX ** 2)
    b = (somaY - a * somaX) / n

    if b == 0:
        print(f"Peter, a equação da reta deve ser y = {a:.1f}X!")
    elif b > 0:
        print(f"Peter, a equação da reta deve ser y = {a:.1f}X + {b:.1f}!")
    else:
        print(f"Peter, a equação da reta deve ser y = {a:.1f}X {b:.1f}!")

n = int(input())
xArray = [int(input()) for _ in range(n)]
yArray = [int(input()) for _ in range(n)]

calcularRetaMmq(n, xArray, yArray)
```

### Métodos de Riemann e dos Trapézios

```python
def funcao(x, exp):
    return eval(exp, {"x": x, "np": np, "math": math})

def metodoTrapezoides(exp, a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = funcao(x, exp)

    return (h / 2) * (y[0] + 2 * np.sum(y[1:-1]) + y[-1])

def somaRiemann(exp, a, b, n=100000):
    h = (b - a) / n
    xMedio = np.linspace(a, b, n + 1)
    yMedio = funcao(xMedio, exp)

    return np.sum(yMedio) * h

def calculoErro(valorExato, valorAproximado):
    return abs((valorExato - valorAproximado) / valorExato) * 100

def resultado(exp, a, b, n):
    valorAproximado = metodoTrapezoides(exp, a, b, n)
    valorExato = somaRiemann(exp, a, b)
    erroRelativo = calculoErro(valorExato, valorAproximado)

    print(f"Valor aproximado por trapezios: {valorAproximado:.4f}")
    print(f"Valor exato por riemann: {valorExato:.4f}")
    print(f"Erro relativo: {erroRelativo:.4f}%")

    if erroRelativo > 1:
        print("O erro eh significativo")
    else:
        print("Boa aproximacao")

exp = input().strip()
a = float(input().strip())
b = float(input().strip())
n = int(input().strip())

resultado(exp, a, b, n)
```

### Rede Elética

```python
def gaussSeidel(A, B, X, maxIteracoes, precisao):
    n = len(A)
    for iteracao in range(maxIteracoes):
        XAntigo = X[:]
        for i in range(n):
            resultado = B[i] 
            for j in range(n):
                if i != j:
                    resultado -= A[i][j] * X[j]
            X[i] = resultado / A[i][i]
        
        if all(abs(X[i] - XAntigo[i]) < precisao for i in range(n)):
            print(f"Solução encontrada após {iteracao + 1} iterações.")
            return X
    
    print(f"Solução encontrada após {maxIteracoes} iterações.")
    return X


A = [list(map(float, input().split())) for i in range(3)]

B = list(map(float, input().split()))

X = list(map(float, input().split()))

maxIteracoes = int(input())

precisao = float(input())

resultado = gaussSeidel(A, B, X, maxIteracoes, precisao)

print(f"Solução aproximada: V1 = {resultado[0]:.4f}, V2 = {resultado[1]:.4f}, V3 = {resultado[2]:.4f}")
```

### Problema do paleontólogo

```python
def jacobi(A, B, maxIteracoes=30, minErro=1e-10, maxErro=1e+4):
    n = len(A)
    X = np.zeros(n)

    for iteracao in range(maxIteracoes):
        novoX = np.zeros(n)

        for i in range(n):
            resultado = B[i] - sum(A[i][j] * X[j] for j in range(n) if j != i)
            novoX[i] = resultado / A[i][i]

        erro = np.linalg.norm(novoX - X, ord=np.inf)

        if erro > maxErro:
            print("O método não convergiu :( Escolha outro lugar por favor!")
            return

        if erro < minErro:
            novoX = np.round(novoX, 1)
            print(f"Começe a escavar em: {novoX}")
            return

        X = novoX

    X = np.round(X, 1)
    print(f"Começe a escavar em: {X}")

n = int(input())

A = []
for i in range(n):
    linha = []
    for j in range(n):
        valor = float(input())
        linha.append(valor)
    A.append(linha)

B = [float(input()) for i in range(n)]

jacobi(A, B)
```

### Encontre o One Piece!

```python
def interpolacaoPorPares(x, fx, op):
    n = len(x)

    if op < x[0] or op > x[-1]:
        print("O One Piece está fora do intervalo. Luffy você disse os valores corretos?")
        return

    for i in range(n - 1):
        if x[i] <= op <= x[i + 1]:
            f_op = fx[i] + (fx[i + 1] - fx[i]) * ((op - x[i]) / (x[i + 1] - x[i]))
            print(f"O One Piece está na posição {round(f_op, 1)}")
            return

n = int(input())
x = [float(input()) for _ in range(n)]
fx = [float(input()) for _ in range(n)]
op = float(input())

interpolacaoPorPares(x, fx, op)
```

### Queda de ações!

```python
def iterpolacaoLagrange(x, y, xPrevisto):
    n = len(x)
    yPrevisto = 0

    for i in range(n):
        L = 1
        for j in range(n):
            if i != j:
                L *= (xPrevisto - x[j]) / (x[i] - x[j])
        yPrevisto += y[i] * L

    return yPrevisto

X = []
Y = []

for i in range(3):
    X.append(int(input()))
for i in range(3):
    Y.append(float(input()))

xPred = 4

precoPrevisto = iterpolacaoLagrange(X, Y, xPred)

print(f"O preço previsto para as ações da NVDA no dia {xPred} é: ${precoPrevisto:.2f}")
```

### Deu a louca no gato de Schrödinger

```python
def iterpolacaoLinear(t, t0, t1, x0, x1):
    return x0 + (x1 - x0) * (t - t0) / (t1 - t0)

def iterpolacaoQuadratica(t, t0, t1, t2, x0, x1, x2):
    A = np.array([[t0**2, t0, 1],
                  [t1**2, t1, 1],
                  [t2**2, t2, 1]])
    B = np.array([x0, x1, x2])
    coeficientes = np.linalg.solve(A, B)
    a, b, c = coeficientes
    return a * t**2 + b * t + c

def calculoIterpolacaoLinear(tempos, posicoes, tIntermediario):
  resultado = 0
  for i in range(len(tempos) - 1):
    if tempos[i] <= tIntermediario <= tempos[i + 1]:
        resultado = iterpolacaoLinear(tIntermediario, tempos[i], tempos[i + 1], posicoes[i], posicoes[i + 1])

  print(f"A posicao do gato (spline linear) eh: {resultado:.4f}")

def calculoIterpolacaoQuadratica(tempos, posicoes, tIntermediario):
  resultado = iterpolacaoQuadratica(tIntermediario, tempos[0], tempos[1], tempos[2], posicoes[0], posicoes[1], posicoes[2])
  print(f"A posicao do gato (spline quadratica) eh: {resultado:.4f}")

tempos = list(map(int, input().split()))
posicoes = list(map(int, input().split()))
tIntermediario = int(input())

calculoIterpolacaoLinear(tempos, posicoes, tIntermediario)
calculoIterpolacaoQuadratica(tempos, posicoes, tIntermediario)
```
